# defaultdic
- Creates a dict-like object whose missing keys are auto-created by calling list() (an empty list) as the default.
- Accessing a missing key creates and returns an empty list: collectionscounter['a'] → [] and the key 'a' now exists.
- Useful for grouping/collecting values under keys without checking if key in d first.

````python
from collections import defaultdict
d = defaultdict(list)
pairs = [('a', 1), ('b', 2), ('a', 3)]
for k, v in pairs:
    d[k].append(v)
# d -> {'a': [1, 3], 'b': [2]}
print(d)
print('------')
print(list(d.keys()))
````

# `namedtuple`

`namedtuple` creates tuple-like classes with named fields.

- Immutable
- Lightweight and memory efficient
- Supports tuple behavior (`indexing`, `unpacking`)
- Also supports attribute access (`obj.field_name`)

`namedtuple` vs `tuple` vs `dict`

| Feature | `tuple` | `namedtuple` | `dict` |
|---|---|---|---|
| Access style | by index | by index + attribute | by key |
| Mutability | immutable | immutable | mutable |
| Fixed structure | yes | yes | no |
| Readability | low/medium | high | high |
| Memory usage | low | low | higher |

Use `namedtuple` when:

- You want simple records with fixed fields
- You need readability without full class boilerplate
- You want immutable data containers

Prefer `dict` when keys may vary dynamically.
Prefer a full class/dataclass when behavior (methods/validation) grows.

## Common Methods and Attributes

Assume:

```python
from collections import namedtuple

Person = namedtuple('Person', ['name', 'age', 'city'])
p = Person('Alice', 30, 'Seattle')
```

### 1. `_fields`
Returns field names as a tuple.

```python
print(Person._fields)
# ('name', 'age', 'city')
```

### 2. `_asdict()`
Returns an ordered dictionary of fields to values.

```python
print(p._asdict())
# {'name': 'Alice', 'age': 30, 'city': 'Seattle'}
```

### 3. `_replace(**kwargs)`
Returns a new instance with selected fields replaced.

```python
p2 = p._replace(age=31)
print(p2)
# Person(name='Alice', age=31, city='Seattle')
```

### 4. `_make(iterable)`
Creates an instance from an iterable.

```python
data = ['Bob', 25, 'Austin']
p3 = Person._make(data)
print(p3)
# Person(name='Bob', age=25, city='Austin')
```

### 5. tuple features still work

```python
name, age, city = p
print(name, age, city)
# Alice 30 Seattle
```
