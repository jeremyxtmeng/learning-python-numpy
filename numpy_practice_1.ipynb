### Q 1

100 candies: you pick one coin and flip it. If you get head, you eat two candies. If you get tail, you eat one candy. What's the expected number of times you need to flip the coin before eating all the candies?

```python
import numpy as np
rng=np.random.default_rng()
rng.uniform(0,1)
# one game
i=0
candies=100
while candies>0:
    i+=1
    eat2=rng.uniform(0,1)>=0.5 # if >0.5 True ->1 False->0
    candies-=(int(eat2)+1)
print(i)
```

```python
# one game def
def one_game(candies:int)->int:
    rng=np.random.default_rng()
    rng.uniform(0,1)
    # one game
    i=0
    while candies>0:
        i+=1
        eat2=rng.uniform(0,1)>=0.5 # if >0.5 True ->1 False->0
        candies-=(int(eat2)+1)
    return i

print(one_game(100))
```

```python
# list of result
x=[]
run=100
for k in range(run):
    m=one_game(100)
    x.append(m)

y=np.array(x)
print(np.mean(y), 'is the expected value')
```
