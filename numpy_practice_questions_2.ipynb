1) Given a:
Compute the total of all elements in `a`. Then compute the total again but keep the result as a 2D array with shape `(1, 1)`.


````python
import numpy as np
a = np.array([3, -1, 4, 0, 2])
np.sum(a)
np.sum(a, keepdims=True)
````

2) Given:
```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
```
Compute the column-wise totals (one total per column). Also compute the row-wise totals (one total per row).


````python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
print(np.sum(a, axis=0, keepdims=True))

print(np.sum(a, axis=1, keepdims=True))

````

3) Given:
