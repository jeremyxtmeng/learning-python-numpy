# Summary of Python programming

## ways to iterate

- string
    - `for i in str:`
- dict
    - `for k, v in dict.items():`
    - `for k in dict.keys():` is the same as `for k in dict:`
    - `for v in dict.values():`
- list
    - `for i in list_a:`
    - `for i in range(len(list_a)):`
    - jointly iterate over two lists: `for i, j in zip(list_a,list_b):`
    - `for idx, v in enumerate(list_a):`
- Pandas series
    - `for i in pandas.series({'day':[1,2,3]}):`
    - `for idx, v in s.items():`
- Pandas dataframe
    - iterate over rows: `for idx, row in df.iterrows():`
    - iterate over columns: `for colname in df.columns:` or `for col_name, col_data in df.items():` 
- Numpy matrix / Numpy array
    - iterate over rows:
        - `for row in arr:`
        - `for i in range(arr.shape[0]): row = arr[i, :]`
    - iterate over columns:
        - `for col in arr.T:`
        - `for j in range(arr.shape[1]): col = arr[:, j]`
