### enumerate() and zip()

```enumerate(<iterable>, start=0)``` takes iterable object and returns the index and values.

```zip()``` combines two iterables as a tuple and stop at the shortest list in the iteration.

[More on splat operator.](https://lucasoshiro.github.io/posts-en/2024-05-11-splat-operator/)

````python
alphabet=['a', 'b','c', 'x', 'y']
add_letters=['e', 'f','g']
# 1:
for i, v in enumerate(alphabet):
    print(f'index value is {i}, and value is {v}')
# 2
for i in zip(alphabet,add_letters):
    print(f'value is {i}')

for x,y in zip(alphabet,add_letters):
    print(f'value is {x} in the first list and {y} the other')
````

````python
# 3 splat operator
zipped=zip(alphabet, add_letters)
print(zipped)
print(type(zipped))

# create a dictionary
dic2=dict(zipped)
print(dic2)

print(*zipped) # prints nothing because zip is a lazy zip which was consumed in the dict()


````

````python
zipped=zip(alphabet, add_letters)
print(dict(zipped))

zipped=zip(alphabet, add_letters)
print(*zipped) # returns tuples
````

### (opt) list generator
similar to a comprehension, generators don't store the list in memory. This is called lazy evaluation, as the value is assigned only when needed.

```yield``` returns a sequence.

````python
# build a list with a comprehension
squares = (i * i for i in range(6))

# iterate the resulting list
for sq in squares:
    print(sq)

def countdown(n):
    while n > 0:
        yield n
        n -= 1

for value in countdown(5):
    print(value)
````

### writing my own function-based context manager

1. Decorator Syntax: apply ```@contextlib.contextmanager``` to a function that uses a ```yield``` statement.
2. Codes before ```yield``` (like an ```__enter__``` method): when the ```with``` statement is first encountered. 
3. Upon seeing ```yield```, running codes in the ```with ... as var:``` clause.
    * (optional) Codes chunk in ```yield <chunk>``` assigns the variable in the ```with ... as var:``` clause
4. Codes after ```yield``` (like an ```__exit__``` method): runs to exit the block (teardown/cleanup).
5. Exception Handling: The code after ```yield``` is guaranteed to execute even if an error occurs within the with block. 

````python
# eg 1
from contextlib import contextmanager
@contextmanager
def my_context():
    print('hello')
    yield 42 
    print('goodbye')

with my_context() as foo:
    print('foo is {}'.format(foo)) # this is run after hello and 42 is assigned to foo
````

````python
# eg 2
from contextlib import contextmanager
import time

@contextmanager

def timer():
  start = time.time()
  # Send control back to the context block
  try: # ensures codes under finally is run even errors in with clause is encountered 
    yield
  finally:
    end = time.time()
    print('Elapsed: {:.2f}s'.format(end - start))

with timer():
  print('This should take approximately 0.25 seconds')
  time.sleep(0.25)
  
````

### closures

Closures are tuples of parameters that functions need to run but are out of the scope of the function.

````python
def foo():
    a=5
    def bar():
        print(a)
    return bar
func=foo()
func()
print(type(func.__closure__))
print(len(func.__closure__))
func.__closure__[0].cell_contents
````

````python
a=5
def foo(a):
    def bar():
        print(a)
    return bar
func=foo(a)

func()
del(a)

func() # a is stored in the closure of bar and remains even though a in the global scope is deleted.
````

### decorator
Modifying the behavior of functions

````python
def multiply(a, b):
    return a * b
def double_args(func):
    # Define a new function that we can modify
    def wrapper(a, b):
        return func(a*2,b*2)
    # Return the new function
    return wrapper

new_multiply = double_args(multiply)
print(new_multiply.__closure__[0].cell_contents) # new_multiply saved the function multiply in the closure of new_multiply

new_multiply(1, 5)
````

````python
# alternative way
def double_args2(func):
    def wrapper2(a, b):
        return func(a*2,b*2)
    return wrapper2

@double_args2
def multiply2(a, b):
    return a * b

multiply2(1, 5)
````

### decorator real-world usages

````python
# eg 1
import time

def time2(func):

    def wrapper(*args, **kwargs):
        # counting time inside the wrapper
        start=time.time()
        result=func(*args,**kwargs)
        duration=time.time()-start
        #print(f'{func.__name__} take {duration} to run')
        print('{} take {} to run'.format(func.__name__, duration))
        return result
    
    return wrapper

def delay(x):
    k=0
    for i in range(x):
        k+=i
    return k

new_delay=time2(delay)

new_delay(10000000)
````

````python
# eg 2
# create cache to store returned values
def memorize(func):
    cache={}
    def wrapper(*args, **kwargs):
        # Define a hashable key for 'kwargs'.    
        kwargs_key = tuple(sorted(kwargs.items()))
        # If these arguments haven't been seen before,
        if (args, kwargs_key) not in cache:
            # Call func() and store the result.
            cache[(args, kwargs_key)] = func(*args, **kwargs)
        return cache[(args, kwargs_key)]
    return wrapper


@memorize
def delay2(x):
    print('slow function')
    k=0
    for i in range(x):
        k+=i
    return k

print(delay2(1000)) # the first time delay is called, it stores wrapper function and a closure cache for wrapper

# here results are printed in the cache so the text is not printed
# the second time the wrapper is call, it uses the stored closure.
print(delay2(1000))
````

````python
# count the appearance of a function

def counter(func):
  def wrapper(*args, **kwargs):
    wrapper.count += 1
    # Call the function being decorated and return the result
    return func(*args, **kwargs)
  # Set count to 0 to initialize call count for each new decorated function
  wrapper.count = 0
  # Return the new decorated function
  return wrapper

# Decorate foo() with the counter() decorator
@counter
def foo():
  print('calling foo()')
  
foo()
foo()

print('foo() was called {} times.'.format(foo.count))
````

### decorator and metadata

The doc string of the decorator returns the doc of the wrapper instead of the function we are decorating. Use build-in decorator  ```wrap``` to solve the problem.

````python
from functools import wraps

def add_hello(func):
  # Decorate wrapper() so that it keeps func()'s metadata
  @wraps(func)
  def wrapper(*args, **kwargs):
    """Print 'hello' and then call the decorated function."""
    print('Hello')
    return func(*args, **kwargs)
  return wrapper
  
@add_hello
def print_sum(a, b):
  """Adds two numbers and prints the sum"""
  print(a + b)
  
print_sum(10, 20)
print_sum_docstring = print_sum.__doc__
print(print_sum_docstring)
````

can use ```__wrapped__``` methods to use the original function

````python
print_sum.__wrapped__(10, 20)
````

### decorator with argument

````python
import numpy as np
def run_n_times(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for i in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

# to show results
# way 1
# function takes three argument 
run_n_times(2)(print)('hello')

# way 2
outer=run_n_times(3)
inner1=outer(print)
inner2=inner1('good morning')
inner2

# way 3
@run_n_times(4)
def print2():
   print('\n good evening')

print2()
````
