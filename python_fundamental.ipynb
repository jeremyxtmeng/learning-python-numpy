````python
# RETURNING A FUNCTION
def grumpy():
    """
    no parameter
    """
    print("I am ok")
    def no_n_times(n):
        """
        n, inner para
        1st inner function
        """
        print("No",n, "times...")
        def no_m_more_times(m):
            """
            m, the most inner para
            2nd inner function
            """
            print("... and no", m, "more times")
            for i in range(n+m):
                print("no")
        return no_m_more_times # Recall that putting parentheses after the function name makes a function call, which you don’t want to do.
    return no_n_times

grumpy()(2)(3)
````

````python
# PASSING FUNCTIONS AS PARAMETERS
# A function sandwich takes in a parameter named
# kind_of_sandwich. This parameter is a function
# object. Inside the sandwich function, you can call
# kind_of_sandwich as usual by adding parentheses
# after it.

# When you call the sandwich function, you call it with a
# function object as a parameter. You give it the name of a
# function for the sandwich you want to make. You don’t
# put parentheses after blt or breakfast as the
# argument because you want to pass the function object
# itself. If you use blt() or breakfast(), this will be a
# string object because this is a function call that returns a
# string

def sandwich(kind_of_sandwich):
    print("--------")
    print(kind_of_sandwich ())
    print("--------")

def blt():
    my_blt = " bacon\nlettuce\n tomato"
    return my_blt

def breakfast():
    my_ec = " eggegg\n cheese"
    return my_ec

print(sandwich(blt))
````

````python
# Mutable objects as function parameters

def add_word(d, word, definition):
    """
    d, dict that maps strings to lists of strings
    word, a string
    definition, a string
    Mutates d by adding the entry word:definition
    If word is already in d, append definition to word's
    value list
    Does not return anything
    """
    if word in d:
        d[word].append(definition)
    else:
        d[word] = [definition]
    return 1

words = {}
k=add_word(words, 'box', 'fight')
print(k)  # returns 1
print(words)
````

- `__init__()` in Python is the initializer method for a class, automatically called when you create a new object (instance) to set up its initial state, assign attribute values (like self.name = name)
> Making an object from a class is called instantiation
- by convention, defining the class using CamelCase followed by docstring
- Any variable prefixed with `self` is available to every method in the class, and we’ll also be able to access these variables through any instance created from the class.
    - `self.xxx` defines the attributes of objects in a class
        - default ones not in `__init__(self, att1, attr2...)`, and can be modified through `object_name.default_one=##`
        - assigned ones are those in `__init__(self, att1, attr2...)`
- making a class: `class_name(attributes1, attribute2)`
- accessing attributes: `object_name.attribute1`
- calling method; `object_name.method_name()`

> WARNING: pay attention to () as **object_name.attribute**= vs **object_name.method()**

````python
class Stack(object):
    def __init__( self):
        self.stack = []
        self.name='I am a list'
    def get_stack_elements(self):
        return self.stack.copy()
    def add_one(self , item):
        self.stack.append(item)
    def add_many(self , item, n):
        for i in range(n):
            self.stack.append(item)
    def remove_one(self):
        self.stack.pop()
    def remove_many(self , n, beginning=None):
        assert beginning==None or beginning==0, "must enter nothing or 0"
        for i in range(n):
            if beginning==0:
                self.stack.pop(beginning)
            else:
                self.stack.pop()
    def size(self):
            return len(self.stack)
    def prettyprint(self):
        for thing in self.stack[::-1]:
            print('|_',thing, '_|')
    # Write a method for the Stack object, named add_list, 
    # which takes in a list as a parameter. Each element in 
    # the list is added to the stack, with items at the 
    # beginning of the list being added to the stack first.
    def add_list(self, L):
        self.stack.extend(L)
    def __str__(self):
        ret = ""
        for thing in self.stack[::-1]:
            ret += "|_ " + str(thing) + " _|\n"
        return ret

X=Stack()
X.add_list([1,2])
print(Stack.get_stack_elements(X))
Stack.prettyprint(X)
X.prettyprint()
````

Summary

Your question
Why does this not work in `Stack.__str__`:
- `return str(self.prettyprint())`
- and why do we need `str(thing)` in:
  `ret += '|_ ' + str(thing) + ' _|\n'`?

Answer
- `prettyprint()` **prints** lines but does not return a string, so it returns `None`.
- `__str__` must **return a string**, not `None`.
- Therefore `str(self.prettyprint())` becomes `"None"` (or behaves unexpectedly).
- `str(thing)` is needed because `thing` may be an object (e.g., `Circle`), and string concatenation requires strings.


Correct pattern
```python
def __str__(self):
    ret = ""
    for thing in self.stack[::-1]:
        ret += "|_ " + str(thing) + " _|\n"
    return ret
```

````python
class Fraction(object):
    '''magic operations on fractions
    Attributes:
        top: numerator
        botton: denominator
    '''
    def __init__(self, top, bottom):
        self.top = top
        self.bottom = bottom

    def __add__(self, other_fraction):
        new_top = self.top * other_fraction.bottom + self.bottom * other_fraction.top
        new_bottom = self.bottom * other_fraction.bottom
        return Fraction(new_top, new_bottom)
    def __sub__(self, other_fraction):
        new_top = self.top - other_fraction.bottom - self.bottom * other_fraction.top
        new_bottom = self.bottom * other_fraction.bottom
        return Fraction(new_top, new_bottom)

    def __mul__(self, other_fraction):
        new_top = self.top * other_fraction.top
        new_bottom = self.bottom * other_fraction.bottom
        return Fraction(new_top, new_bottom)
    # change the behavior of print
    def __str__(self): 
        return str(self.top)+"\n--\n"+str(self.bottom)

pokemon=Fraction(1,2)
pokemon.__doc__
half = Fraction(1,2)
quarter = Fraction(1,4)
print(half - quarter)
````
