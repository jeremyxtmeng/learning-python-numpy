## uniform distribution

## normal

* Ex1: explain normal distribution to a production manager
    * We launch a new product. Our site has 1000 new users **independently** making purchase decision with a probability $p$ around 5\%. We calculate the # of purchases /1000. We collect this data over time and plot it. The curve look approximately like a normal distribution. 
* Ex2: explain the central limit theorem. 
    * The growth of daily active users stalled for our site. We want to know the average age of our users. Each day we randomly sample 1000 users (or a number large enough) and calculate the mean age. I do this over and over again. The plot of this data follows a normal distribution. The average of averages approximate population mean and the std of the average approximate the population std adjusted by dividing the square root of daily sample size.

```python
from scipy.stats import norm

print(norm.cdf(1.98, 0, 1)) # mean 0 std 1

print(norm.ppf(0.95, 0, 1))
```

## possion
The prob of arrival is fixed.

Note: the standard score z=(obs-mean)/std

```python
# the average number of events in an interval lambda (also the expected value)
# what is the prob of # <5
from scipy.stats import poisson
print(poisson.pmf(5, 8)) # the prob of 5 events when the mean value of events in the time interval is 8
print(poisson.cdf(5, 8)) # the prob of less than 5 events when the mean value of events in the time interval is 8

```

### Poisson and exponential

In terms of rate (Poisson):
* $\lambda=0.5$ trains per minute

In terms of time between events (exponential):
* $1/\lambda=2$ means 1 train per 2 minutes
    


```python
from scipy.stats import expon

print(f'what is the wait time greater than 4 minutes is P(wait>4): {1-expon.cdf(4, scale=2)}')
```
